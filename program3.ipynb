```python
import matplotlib.pyplot as plt
import numpy as np
xpoints=np.array([2,13])
ypoints=np.array([5,9])
plt.subplot(2,3,1)
plt.title("title1")
plt.xlabel("orange",color="blue",loc='right')
plt.ylabel("apple",color="blue")
plt.plot(xpoints,ypoints,'o')

#2graph
xpoints=np.array([2,5,2,5])
ypoints=np.array([5,2,5,2])
plt.subplot(2,3,2)
plt.plot(xpoints,'--')
plt.plot(ypoints,marker='*')
#3graph
xpoints=np.array([2,5,2,9])
plt.subplot(2,3,3)
plt.plot(xpoints,marker='o',ms=10,mec='black',mfc='yellow')
#4graph
xpoints=([0,40])
ypoints=([0,40])
plt.subplot(2,3,4)
plt.title("title2")
plt.grid()
plt.plot(xpoints,ypoints)
#5graph
x=np.array([2,4,5,7,9,13,18])
y=np.array([0,1,4,6,7,3,15,])
plt.subplot(2,3,5)
colors=np.array(["red","green","yellow","black","pink","hotpink","purple"])
plt.scatter(x,y,c=colors,alpha=0.5)

#6graph
x=np.array(["a","b","c"])
y=np.array([6,4,12])
plt.subplot(2,3,6)
plt.bar(x,y,color=["red","green","yellow"],width=0.1)
plt.tight_layout()
plt.suptitle("GRAPH")
plt.show()
```

```python
#pichart
import matplotlib.pyplot as plt
import numpy as np
y=np.array([35,20,20,25])
mylabels=["apple","orange","grape","banana"]
myexplode=[0,2,0,0]
plt.pie(y,labels=mylabels,startangle=90,explode=myexplode,shadow=True)
plt.legend()
plt.show()
```
