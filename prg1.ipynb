```python
#largest number from the list
numbers=input("Enter the numbers:").split(",")
largest=numbers[0]
for num in numbers:
       if num>largest:
            largest=num
print("largest number is:",largest)
```

```python
#removing duplicates
numb=input("Enter the numbers:").split(",")
unique=[]
for num in numb:
    if num not in unique:
        unique.append(num)
print("List after removing duplicate:")
print(unique)
```

```python
#tuple to dictionary
n=int(input("Enter the number of key_value pairs:"))
pairs=[]
for i in range(n):
    key=input("Enter key:")
    value=input("Enter value:")
    pairs.append((key,value))
my_dict=dict(pairs)
print("dictionary:")
print(my_dict)
```

```python
#mmerge 2 dict

Rno = input("Enter the roll numbers for 3 students (using space): ").split()
Rno1 = list(map(int, Rno))
names = input("Enter the names of 3 students (using space): ").split()

students = dict(zip(Rno1, names))
print("Students with roll numbers:", students)
Rno2 = input("Enter the roll numbers for 2 students (using space): ").split()
Rno3 = list(map(int, Rno2))

names1 = input("Enter the names of 2 students (using space): ").split()

students1 = dict(zip(Rno3, names1))
print("Students with roll numbers:", students1)
students.update(students1)

print("Merged dictionary:", students)
```

```python
list1=list(map(int,input("Enter the first list:").split()))
list2=list(map(int,input("Enter the second list:").split()))
found=False
for item in list1:
    if item in list2:
        found=True
        break
if found:
    print("the list have one common member")
else:
    print("the list do not have common member")
```
